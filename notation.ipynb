```python
historique_notes = []

print("Entre tes notes et leurs coefficients. Tape 'fin' pour arrêter.")

while True:
    entree = input("Note : ")
    
    if entree.lower() == 'fin':
        break
        
    try:#fonctionne avec except, lance la boucle except si une erreur survient dans le try
        note = float(entree)
        coef = float(input("Coefficient : "))
        
        if coef <= 0:
            print("Le coefficient doit être supérieur à 0.")
            continue
            
        # Stockage (note, coef)
        historique_notes.append((note, coef))
    except ValueError:
        print("Saisie invalide. Merci d'entrer des nombres réels.")

# Calcul de la moyenne à partir des données stockées
if historique_notes:
    somme= sum(note * coef for note, coef in historique_notes)
    somme_coefficients = sum(coef for _, coef in historique_notes)
    
    moyenne = somme/ somme_coefficients
    
    print(f"\nNombre de notes saisies : {len(historique_notes)}")
    print(f"Ta moyenne générale est de : {moyenne:.2f}")
else:
    print("Aucune note valide n'a été saisie.")
if moyenne >10 :
    print("bravo tu passe en 2eme année")
else : 
    print("moyenne non sufisante")
    
    
```

```python
historique_notes = []

print("Entre tes notes et leurs coefficients. Tape 'fin' pour arrêter.")

while True:
    entree = input("Note : ")
    
    if entree.lower() == 'fin':
        break
        
    try:
        note = float(entree)
        coef = float(input("Coefficient : "))
        
        if coef <= 0:
            print("Le coefficient doit être supérieur à 0.")
            continue
            
        # Stockage sous forme de dictionnaire {clé: valeur}
        historique_notes.append({"note": note, "coef": coef})
    except ValueError:
        print("Saisie invalide. Merci d'entrer des nombres réels.")

# Calcul de la moyenne à partir des dictionnaires
if historique_notes:
    somme= sum(dico["note"] * dico["coef"] for dico in historique_notes)
    somme_coefficients = sum(dico["coef"] for dico in historique_notes)
    
    moyenne = somme / somme_coefficients
    
    print(f"\nNombre de notes saisies : {len(historique_notes)}")
    print(f"Ta moyenne générale est de : {moyenne:.2f}")
else:
    print("Aucune note valide n'a été saisie.")
if moyenne >10 :
    print("bravo tu passe en 2eme année")
else : 
    print("moyenne non sufisante")
```
